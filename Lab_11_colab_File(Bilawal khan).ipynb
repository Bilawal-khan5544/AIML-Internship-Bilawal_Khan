## A Real Feature Vector to Work With

We'll use one real sample from the Wine dataset (built into scikit-learn, no download needed) so the numbers in this lab mean something real, even though we aren't training a model yet.

```
from sklearn.datasets import load_wine
import numpy as np

wine = load_wine()
X = wine.data
y = wine.target

print("Dataset shape:", X.shape)
first_sample = X[0]
print("First sample (13 features):", first_sample)
print("First sample's true class:", y[0])
```

## Example 1: A Single Neuron, Computed by Hand

```
def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# A single neuron with 3 made-up inputs (for simplicity, not the full 13-feature sample yet)
x = np.array([1.5, 2.0, 0.5])
w = np.array([0.4, -0.2, 0.9])
b = 0.1

z = np.dot(x, w) + b
output = relu(z)

print("Weighted sum (z):", z)
print("Neuron output (after ReLU):", output)
```

## Example 2: Manual Forward Propagation (Two Layers)

```
X_toy = np.array([1.5, 2.0, 0.5])

W1 = np.array([[0.2, -0.5],
               [0.4,  0.1],
               [-0.3, 0.8]])
b1 = np.array([0.1, -0.2])

W2 = np.array([0.6, -0.9])
b2 = 0.05

hidden_z = np.dot(X_toy, W1) + b1
hidden_a = relu(hidden_z)

output_z = np.dot(hidden_a, W2) + b2
output = sigmoid(output_z)

print("Hidden layer activations:", hidden_a)
print("Final output (probability):", output)
```

## Example 3: Calculating Loss

```
y_true = 1

mse = (y_true - output) ** 2
bce = -(y_true * np.log(output) + (1 - y_true) * np.log(1 - output))

print("Mean Squared Error:", mse)
print("Binary Cross-Entropy:", bce)
```

## Example 4: One Gradient Descent Step, by Hand

```
w = 0.5
x_val, y_true_val = 2.0, 3.0
learning_rate = 0.1

prediction = w * x_val
loss = (prediction - y_true_val) ** 2

gradient = 2 * (prediction - y_true_val) * x_val
w_new = w - learning_rate * gradient

print("Prediction before update:", prediction)
print("Loss before update:", loss)
print("Gradient:", gradient)
print("Updated weight:", w_new)
```

```
# Task 1: 3-layer forward pass (13 -> 4 -> 3 -> 1) on a real wine sample
np.random.seed(1)
sample = X[10]
sample = (sample - X.mean(axis=0)) / X.std(axis=0)   # standardize so values are on a similar scale

W1 = np.random.randn(13, 4) * 0.5;  b1 = np.full(4, 0.1)
W2 = np.random.randn(4, 3) * 0.5;   b2 = np.full(3, 0.1)
W3 = np.random.randn(3) * 0.5;      b3 = 0.0

a1 = relu(np.dot(sample, W1) + b1)
a2 = relu(np.dot(a1, W2) + b2)
out = sigmoid(np.dot(a2, W3) + b3)

print("Hidden layer 1 (4 neurons):", a1)
print("Hidden layer 2 (3 neurons):", a2)
print("Final output (probability):", out)
```

```
# Task 2: Compare activation functions on the same z
z_test = 0.8
print("ReLU   :", relu(z_test))
print("Sigmoid:", sigmoid(z_test))
print("Tanh   :", np.tanh(z_test))

# Bonus: a negative z shows the difference more clearly
z_neg = -0.8
print("\nWith z = -0.8:")
print("ReLU   :", relu(z_neg))
print("Sigmoid:", sigmoid(z_neg))
print("Tanh   :", np.tanh(z_neg))
```

**Task 2 comparison:** For the same `z = 0.8`, ReLU passes the value through unchanged (0.8), Sigmoid squashes it into the range (0, 1) giving about 0.690, and Tanh squashes it into (-1, 1) giving about 0.664. With a negative `z` (-0.8), ReLU outputs 0, Sigmoid gives about 0.31, and Tanh gives about -0.664, so Tanh keeps the sign while ReLU discards negatives. Sigmoid suits output probabilities, ReLU suits hidden layers, and Tanh is zero-centered.

```
# Task 3: MSE vs Binary Cross-Entropy
y_true = 1
for pred in [0.9, 0.5, 0.1]:
    mse = (y_true - pred) ** 2
    bce = -(y_true * np.log(pred) + (1 - y_true) * np.log(1 - pred))
    print(f"Prediction {pred}: MSE = {mse:.4f}, BCE = {bce:.4f}")
```

**Task 3 comparison:** As the prediction moves away from the true label 1 (0.9 → 0.5 → 0.1), MSE grows modestly (0.01 → 0.25 → 0.81) and is capped at 1. Binary Cross-Entropy grows much faster (0.105 → 0.693 → 2.303) and goes to infinity as the prediction approaches 0. So BCE punishes confident wrong predictions far more harshly, which gives stronger gradients for classification.

```
# Task 4: 20 steps of gradient descent + plot
import matplotlib.pyplot as plt

w = 0.5
x_val, y_true_val = 2.0, 3.0
learning_rate = 0.1
losses = []

for step in range(20):
    prediction = w * x_val
    loss = (prediction - y_true_val) ** 2
    losses.append(loss)
    gradient = 2 * (prediction - y_true_val) * x_val
    w = w - learning_rate * gradient

plt.plot(range(1, 21), losses, marker='o')
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Loss over 20 gradient descent steps")
plt.grid(True)
plt.show()

print("Final weight:", w, "(ideal = 1.5)")
print("Loss by step:", [round(l, 6) for l in losses])
```

**Task 4 observation:** Yes, the loss flattens out very quickly: it drops from 4.0 to 0.16, 0.0064, 0.000256 and about 0.00001 by step 5. After roughly step 5 it is essentially zero (the weight converges to 1.5), so the remaining steps add almost nothing.
