## Getting the Dataset

This lab uses the Tips dataset, built directly into Seaborn — no download or upload needed.

```python
import seaborn as sns
import pandas as pd
import numpy as np

df = sns.load_dataset("tips")
print(df.shape)
df.head()
```

## Example 1: Encoding Categorical Variables

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
df["day_encoded"] = encoder.fit_transform(df["day"])
print(df[["day", "day_encoded"]].drop_duplicates())
```

```python
df_encoded = pd.get_dummies(df, columns=["day", "time"], drop_first=True)
print(df_encoded.columns.tolist())
df_encoded.head()
```

## Example 2: Creating a New Feature

```python
df["tip_percentage"] = (df["tip"] / df["total_bill"]) * 100
df[["total_bill", "tip", "tip_percentage"]].head()
```

## Example 3: Binning a Continuous Variable

```python
df["bill_category"] = pd.cut(
    df["total_bill"],
    bins=[0, 15, 30, 100],
    labels=["Low", "Medium", "High"]
)
print(df["bill_category"].value_counts())
```

## Example 4: Scaling (StandardScaler vs. MinMaxScaler)

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

standard_scaler = StandardScaler()
minmax_scaler = MinMaxScaler()

df["total_bill_standard"] = standard_scaler.fit_transform(df[["total_bill"]])
df["total_bill_minmax"] = minmax_scaler.fit_transform(df[["total_bill"]])

df[["total_bill", "total_bill_standard", "total_bill_minmax"]].head()
```

## Example 5: Correlation for Feature Selection

```python
correlation = df.select_dtypes(include=[np.number]).corr()
print(correlation["tip"].sort_values(ascending=False))
```

```python
# Task 2: two new derived features
df["is_weekend"] = df["day"].isin(["Sat", "Sun"]).astype(int)   # weekend flag
# bill_per_person already created in Practice Exercise 2
print(df[["day", "is_weekend", "total_bill", "size", "bill_per_person"]].head())
print(df.groupby("is_weekend")["tip"].mean())
```

**Task 2 reasoning**

- `bill_per_person` = `total_bill / size`: normalizes the bill by party size, so a $60 bill for 6 people is distinguishable from $60 for 2. Spending per diner is a better indicator of how generous a tip might be than the raw total.
- `is_weekend` (Sat/Sun = 1): the 4-level `day` column collapses into one binary flag that reflects a plausible behavioural difference (weekend dining is more leisurely and often more celebratory), with fewer columns than one-hot encoding `day`.

```python
# Task 3: bin party size
df["size_category"] = pd.cut(df["size"], bins=[0, 2, 4, 10], labels=["Small", "Medium", "Large"])
print(df["size_category"].value_counts())
```

```python
# Task 4: correlation of numeric features with tip
# Excluded: tip_percentage (computed from tip, so it leaks the target) and the
# scaled copies of total_bill (identical correlation, redundant).
drop = ["tip_percentage", "total_bill_standard", "total_bill_minmax"]
num = df.select_dtypes(include=[np.number]).drop(columns=drop)
corr_tip = num.corr()["tip"].drop("tip").sort_values(ascending=False)
print(corr_tip)
print("\nTop 3:", corr_tip.head(3).index.tolist())
```

**Task 4 explanation**

1. `total_bill` (r ≈ 0.68): tips are usually a percentage of the bill, so a larger bill mechanically means a larger tip. It is by far the strongest predictor.
2. `size` (r ≈ 0.49): bigger parties order more, which raises the bill. Some restaurants also add gratuity or tip more for larger groups.
3. `bill_per_person` (r ≈ 0.35): captures how much each diner spent. It is weaker than `total_bill` because dividing by size removes the party-size effect, but it still signals higher-end orders.

`tip_percentage` was left out on purpose: it is computed from `tip`, so using it as a predictor would be target leakage.

## Task 5: Reflection

Of the techniques in this lab, creating derived features (such as `bill_per_person`) would likely help a model most. The raw numeric columns already carry most of the signal, so scaling or binning mainly changes representation rather than adding information, and scaling matters little for tree-based models. A ratio like bill per person captures a relationship the model would otherwise have to learn from two columns. Log-transforming the right-skewed `tip` target can also help linear models, whose errors are better behaved on a less skewed target. One-hot encoding is required for most models, but the categorical columns here (sex, smoker, day, time) have weak correlation with tip, so it adds less predictive value.
